##
While Loop in Python

```python
count = 1

while count < 10:
    print(count,"Hello")
    count = count + 1
print(count)
```

##
 Break and Continue Statements in Python

```python
sum = 0
for i in range(10):
    print(i,"is processing")
    if i % 2 == 0:
        continue
    sum+=i
    print(i,"is added to sum")
print(sum)

```

```python
accuracy = 95
for i in range(100):
    accuracy+=1
    print(accuracy)
    if accuracy == 100:
        break
```

Problem 1 - Min and Max

```python
inp = input()
num = inp.split()

a = int(num[0]); b = int(num[1]); c = int(num[2])

mx = a
mn = a
if b < mn:
    mn = b
if mn > c:
    mn = c

if b > mx:
    mx = b
if c > mx:
    mx = c

print(mn,mx)
```

Problem 2 - C - Even, Odd, Positive and Negative

```python
n = int(input())
inp = input()
num = inp.split()

positive = 0
negative = 0
even = 0
odd = 0

for i in range(n):
    x = int(num[i])

    if x > 0:
        positive+=1
    elif x < 0:
        negative+=1

    if x % 2 == 0:
        even+=1
    else:
        odd+=1
print("Even: ", even)
print("Odd: ",odd)
print("Positive: ",positive)
print("Negative: ",negative)
```

Problem 3 - Digits

```python
t = int(input())

for i in range(t):
    number = int(input())
    if number == 0:
        print(0)
        continue
    while number > 0:
        print(number%10, end = " ")
        number//=10 ## number = int(number/10)
    print()
```
