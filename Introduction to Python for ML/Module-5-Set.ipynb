```python
S = {10,1,2,3,1,3,4,5}
print(S) # Unordered

S.add(100)
print(S)

S.pop()
print(S)

S.remove(100)
print(S)
```

Union and Intersection

```python
A = {1,2,3,4}
B = {2,5,6,7}
print(A.union(B))
print(A.intersection(B))
print(A.isdisjoint(B))
print(A.issubset(B))
```
