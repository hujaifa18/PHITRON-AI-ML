Access

```python
tup = (10,2,3,2.3,'string',True)
print(tup[5])

# Slicing
new = tup[0:3]
print(new)
```

Immutability

```python
# list->Mutable
lst = [10,20,30]
lst.append(40)
lst[0] = 100
print(lst)

#Tuple->Immutable
tup = (10,20,20,40)
tup[0] =100
print(tup)
```
