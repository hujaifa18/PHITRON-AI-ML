List as Queue

```python
queue = []

# Push in queue
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

# removing front
queue.pop(0)

#get front element
print(f"front:{queue[0]}")

# removing front
queue.pop(0)

#get front element
print(f"front:{queue[0]}")
```

List Comprehension

```python
even = []

for i in range(21):
    if i%2==0:
        even.append(i)
print(even)

random = [x for x in range(10)]
print(random)

random2 = [x*2 for x in range(10)]
print(random2)

random_even =[x for x in range(21) if x%2==0]
print(random_even)
```

List Comprehension with List of String

```python
fruits = ['licchi','apple','orange']
capitalized = []

for fruit in fruits:
    capitalized.append(fruit.capitalize())
print(capitalized)

students = ['rabbi','hujaifa','shishir']
upper_case = []

for student in students:
    upper_case.append(student.upper())
print(upper_case)


names = ['rakib','shihab','imran']
names_upper = [name.upper() for name in names]
print(names_upper)
```
