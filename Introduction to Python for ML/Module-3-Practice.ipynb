E. Lowest Number

```python
import sys
n = int(input())
nums = []
inp = input()
a = inp.split()
for i in range(n):
    nums.append(int(a[i]))

mn = sys.maxsize
idx = -1
for i in range(n):
    if mn > nums[i]:
        mn = nums[i]
        idx = i+1
print(mn,idx)
```

M. Replace MinMax

```python
import sys
n = int(input())
nums = []
inp = input()
a = inp.split()
for i in range(n):
    nums.append(int(a[i]))

mn = sys.maxsize
mn_idx = -1
mx = -sys.maxsize-1
mx_idx = -1
for i in range(n):
    if nums[i] < mn:
        mn = nums[i]
        mn_idx = i
    if nums[i] > mx:
        mx = nums[i]
        mx_idx = i

nums[mn_idx] = mx
nums[mx_idx] = mn
for i in range(n):
    print(nums[i],end=" ")
```

K. Sum Digits

```python
n = int(input())
num = input()

ans = 0
for digit in num:
    ans+=int(digit)

print(ans)
```
