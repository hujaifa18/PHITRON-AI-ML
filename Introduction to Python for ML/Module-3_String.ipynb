String Methods

```python
string = "Welcome to Phitron, Phitron ML , ML, ML"

print(len(string))

#check word
print("Phitron" in string)

upper = string.upper()
print(upper)
lower = string.lower()
print(lower)

capital_first = string.capitalize()
print(capital_first)

#starting idx of substring
#if two simmilar word show first word
index = string.find("Phitron")
print(index)
# from last
last = string.rfind("Phitron")
print(last)

# count word
count = string.count("Phitron")
print(count)

# Replace a substring
new_string = string.replace("ML","AI/ML")
print(new_string)
```

String Splitting and Joining 

```python
prompt = "What-is-Phitron,?"

tokens = prompt.split("-") # by default space
print(tokens)

sentance = "-".join(tokens)
print(sentance)

print(type(tokens[0]),type(tokens), type(sentance))
```

Formated String

```python
name = "Abid"
age = 23
height = 5.123000

print(f"His name is {name}. He is {age} years old. His height is {height:.3}")

info = f"His name is {name.upper()}. He is {age} years old. His height is {height:.3}"
print(info)

accuracy = 0.8333
print(f"Model accuracy = {accuracy:.2%}") 
# Make percentage by taking 2 digit after decimal
```
