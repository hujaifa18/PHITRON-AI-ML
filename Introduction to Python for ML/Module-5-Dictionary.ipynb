Dictionary Comprehension

```python
square = {x : x**2 for x in range(10)}
print(square)

even = { x for x in range(11) if x % 2 == 0}
print(even)
```

```python
Co_Ordinates = [(10,20),(2,3),(7,9)]
Locations = ["dhaka","rajshahi","cumilla"]

dic = {co_or : loc for co_or, loc in zip(Co_Ordinates,Locations)}
print(dic)
```

2.Given a string, print frequency of words

```python
string = """A string is a fundamental data type in computer programming used to 
represent and store text. It consists of a sequential character array, which can 
include letters, numbers, punctuation marks, and spaces, typically enclosed within
single or double quotation marks to distinguish it from executable code. Programs 
treat strings as ordered collections, meaning each individual character occupies a
specific position (index) that allows software to access, search, or alter specific 
parts of the text. Developers frequently manipulate strings through common operations 
like concatenation—which links multiple strings together—as well as measuring length, 
converting cases, and slicing text into smaller segments.
"""

words = string.split()
count = {}

for word in words:
    count[word] = count.get(word,0) + 1
print(count)

for k,v in count.items():
    print(f"count of '{k}' is {v}")
```

J. Count Letters

```python
string = input()
cnt = {}
x = len(string)
for i in range(x):
    cnt[string[i]] = cnt.get(string[i],0) + 1
for k,v in sorted(cnt.items()):
    print(f"{k} : {v}")
```
