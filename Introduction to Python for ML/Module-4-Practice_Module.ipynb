Smart Light Pilot

```python
inp = input()
nums = inp.split()
brightness = float(nums[0])
thershold = float(nums[1])
if brightness >= thershold:
    print("ON")
else:
    print("OFF")
```

Sensor Calibrator — Min-Max Scaling

```python
inp = input()
nums = inp.split()
x = float(nums[0])
min_v = float(nums[1])
max_v = float(nums[2])
norm = (x - min_v) / (max_v - min_v)
print(f"{norm:.2f}")
```

Mini Loss Tracker

```python
n = int(input())
tar = float(input())

total = 0
for i in range(n):
    x = float(input())
    total = total + x
avg = total / n

if avg <= tar:
    print("PASS")
else:
    print("RETRY")
```

Majority Vote Gate

```python
n = int(input())
cnt = 0;
for i in range(n):
    words = input()
    if words == 'YES':
        cnt = cnt + 1
if cnt >= n-cnt:
    print("ACCEPT")
else:
    print("REJECT")
```

AI Mood Detector

```python
s = input()
if 'happy' in s or 'joy' in s or 'smile' in s:
    print("Happy Mood")
elif 'sad' in s or 'cry' in s or 'angry' in s:
    print("Sad Mood")
else:
    print("Neutral Mood")
```

AI Brightness Analyzer

```python
ls = input().split()
sum = 0; cnt = 0
for d in ls:
    sum = sum + int(d)
    cnt = cnt + 1
avg = sum/cnt
if avg < 85:
    print("Dark Image")
elif 85 <= avg and avg <= 170:
    print("Normal Image")
else:
    print("Bright Image")
```

AI Bias Detector

```python
ls = input().split()
cntA = 0; cntB = 0
n = 0
for letter in ls:
    n = n + 1
    if letter == 'A':
        cntA = cntA + 1
    elif letter == 'B':
        cntB = cntB + 1
percntage1 = (cntA * 100) / n
percntage2 = (cntB * 100) / n

if percntage1 > 70  or percntage2 > 70:
    print("Biased Model")
else:
    print("Fair Model")
```

AI Keyword Counter

```python
ls = input().lower().split()

cnt = 0

for l in ls:
    if l == 'ai' or l == 'data' or l == 'model' or l == 'train' or l == 'neural' or l == 'learn':
        cnt = cnt + 1

if cnt >= 2:
    print("AI Detected")
else:
    print("Not AI Related")
```
