```python
x = 20 ; y = 9

sum = x+y #summation
sub = x-y #subtraction
div = round(x/y,2) #division 2 unit after decimal
mul = x*y #multiplication
mod = x%y #reminder

print(sum,sub,div,mul,mod)
```

```python
x = 20 ; y = 9

greater_than = x > y
greater_than_equal = 10>=10
print(greater_than,greater_than_equal)

less_than = x < y
less_than_equal = 10<=10
print(less_than,less_than_equal)

equal = x == y
not_equal = x != y
print(equal,not_equal)
```
