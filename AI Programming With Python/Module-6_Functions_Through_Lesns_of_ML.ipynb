Different Types of Argumrnts

```python
def add(a,b,c):  # Positional Arguments
    print(f"{a},{b},{c}")

add(20,30,10)
```

```python
def add(*args):  # Variable Sized Arguments
    print(args)
    for i in args:
        print(i)

add(20,30,10,50,60)
```

KeyWord Arguments

```python
def add(a,b,c):  
    print(f"{a},{b},{c}")

add(b = 20,c = 30,a = 10)
```

Variable Sized KeyWord Arguments

```python
def add(**kwargs):
    print(kwargs)
    for k,v in kwargs.items():
        print(f"{k} : {v}")

add(a = 10,b = 30,c = 20)
```

Return Type

```python
def sum(a,b):
    ans = a + b
    return ans
m = sum(10,20) * 10
print(m)
```

Generator


```python
lst = [x for x in range(500)]

def data_loader(chunk_size,lst):
    for i in range(len(lst)):
        yield lst[i:i+chunk_size]

x = data_loader(5,lst)
print(x,type(x))
print(next(x))
```

```python
lst = [x for x in range(500)]

def loader(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i : i + chunk_size]
x = loader(5,lst)
print(next(x))
```

Lambda Function

```python
##Python lambda functions are small, single-expression anonymous functions created
#  without a name using the lambda keyword (Syntax: lambda arguments: expression) 
# that automatically return the computed result

import pandas as pd

df  = pd.DataFrame({'temperature (F) ': [32,212,98.6,68]})

df['temperature (C) '] = df['temperature (F) '].apply(lambda x: (x-32) * 5/9)
df
```

```python
import pandas as pd

df = pd.DataFrame({'A' : [1,2,3,4], 'B' : [5,6,7,8]})

df['A'] = df['A'].map(lambda x: x**0.5)
df
```

```python
import pandas as pd

df = pd.DataFrame({'total_spent' : [120,600,450,800]})

df['is_high_value'] = df['total_spent'].apply(lambda x : 1 if x>500 else 0)
df
```
