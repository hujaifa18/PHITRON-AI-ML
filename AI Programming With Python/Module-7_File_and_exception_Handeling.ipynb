#1.Print text of sample.txt
#2.Print number of lines
#3.print number of words
#4.Print Number of character
#5.Store in counter_string.txt

```python
from functools import reduce

with open("./sample_data/sample.txt","r") as file:

  #Print all
  content = file.read()
  print(content)

  # Go to 0 index and print number of lines
  file.seek(0)
  lines = file.readlines()
  print(lines)
  total_lines = len(lines)
  print(total_lines)

  #print list of word in each line
  words_in_line = list(map(lambda x: len(x.split()), lines))
  print(words_in_line)
  # print total word
  total_words = reduce(lambda x,y : x+y, words_in_line)
  print(total_words)

  #Cleaning

  #1. Deleting \n
  lines = list(map(str.strip,lines))

  #2. Deleting spaces
  lines = list(map(lambda x: x.replace(" ", ""), lines))

  #3. print total character
  no_of_char_each_line = list(map(lambda x : len(x) ,lines))
  #print(no_of_char_each_line)

  all_char = reduce(lambda x,y:x+y , no_of_char_each_line)
  print(f"no_of_char : {all_char}")


# Store
with open("./sample_data/counter_string.txt","w") as file:
  file.write(f"Print number of lines : {total_lines}\nPrint number of words : {total_words}\nPrint Number of character : {all_char}")
```

#Problem - 1

```python
def file(filename,content):
  with open(filename,"w+") as f:
    f.write(content)
    f.seek(0)
    print(f.read())
file("notes.txt","Hello, this is my first file.")
```

#Problem - 2

```python
def lines(filename):
  with open(filename,"r") as f:
    lst = f.readlines()
    for i in range(len(lst)):
      print(f"{i+1} : {lst[i]}")

lines("data.txt")
```

```python
def position(filename,pos):
  with open(filename,"r") as file:
    file.seek(pos)
    print(file.read())
position("notes.txt",7)
```

#Problem - 5

```python
def read(filename):
  try:
    with open(filename,"r") as file:
      print(file.read())
  except Exception as e:
    print(e)
  finally:
    print("Operation Complete")

read("notes.txt")
```
