```python
inp = input().split()
lst = list(map(int,inp))
ans = [x for x in lst if x%4 == 0 and x%8 != 0]
print(ans)
```

# Question - 2

```python
inp = input().split()
x = int(inp[0])
y = int(inp[1])
def  compute_ratio(x, y):
  try:
    return (x + y) / (x * y)
  except Exception as e:
    print(e)
compute_ratio(x,y)
```

# Question - 5

```python
string = input()
n = len(string)
i = 0; j = n-1
palindrome = True
while i < j:
  if string[i].lower() != string[j].lower():
    palindrome = False
  i+=1
  j-=1
if palindrome == True:
  print("Palindrome")
else:
  print("Not Palindrome")
```
