Operators in Python

```python
x  = 10; y = 3

sum = x + y #Summation
sub = x - y #Subtraction
mul = x * y #Multiplication
div = round(x / y,2) #Division
mod = x % y #Reminder
print(sum,sub,mul,div,mod)

greater_than = x > y
greater_than_equal = x >= y
print(greater_than,greater_than_equal)

less_than = x < y
less_than_equal = x <= y
print(less_than,less_than_equal)

equal = x == y
not_equal = x != y
print(equal,not_equal)
```

Logical Operator

```python
x = 10; y = 5; z = 8

And = (x > y) and (x > z) # All condition TRUE = TRUE else FALSE
print(And)

Or = (x < y) or (x > z) # If 1 condition TRUE = TRUE else FALSE
print(Or)
```

Precendence and Associativity

```python
eq = 10 + 10 / 2 - 5 * 2
print(eq)

# Higher --> ( * , / ) left to right
# Lower --> ( + , - ) left to right
```
